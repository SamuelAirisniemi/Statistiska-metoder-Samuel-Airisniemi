```python
import numpy as np
from linear_regression import LinearRegression

data = np.genfromtxt("C:/Users/samue/Documents/code/Statistiska-metoder-Samuel-Airisniemi/Labb/housing.csv", delimiter=",", dtype=str, skip_header=1)

data = data[data[:, 4] != ""]

X = data[:, 0:8].astype(float)
y = data[:, 8].astype(float)

model = LinearRegression()
model.fit(X, y)

print("b coefficients:", model.b)
print("n samples:", model.n)
print("d features:", model.d)

print("Sample variance:", model.sample_variance(X, y))
print("Standard deviation:", model.standard_deviation(X, y))
print("RMSE:", model.rmse(X, y))
```
